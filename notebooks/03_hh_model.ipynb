```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.0f}".format)

df = pd.read_parquet('../data/vacancies.parquet')

print(df.shape)
print(df.columns)
df.head()
```

```python
has_salary = df["salary_from"].notna() | df['salary_to'].notna()

s = df[has_salary]
print(len(s))
```

```python
print(s["currency"].value_counts(dropna=False))
print(s["salary_period"].value_counts(dropna=False))
```

```python
is_rub = s["currency"] == "RUR"
is_month = s["salary_period"] == "month"

s = s[is_rub & is_month]
print(len(s))
```

```python
target_salary = s[['salary_from', 'salary_to']]
s['target_salary'] = target_salary.mean(axis=1)
s['target_salary'].describe()
```

```python
s[['title', 'salary_from', 'salary_to', 'target_salary']].sort_values(ascending=False, by='target_salary').head(10)
```

```python
s[['title', 'salary_from', 'salary_to', 'target_salary']].sort_values(ascending=True, by='target_salary').head(10)
```

```python
IT_WORDS = [
    "разработ", "программист", "developer", "engineer", "инженер", "devops", "sre",
    r"\bqa\b", "тестиров", "tester", "аналитик", "analyst", "data", r"\bml\b",
    "1с", "1c", "dba", "баз данных", "системный администратор", "sysadmin",
    "архитектор", "frontend", "backend", "fullstack", "python", "java", "golang",
    "php", "android", "ios", "сетев", "network", r"техническ\w* поддерж",
    "безопасност", "security", "QA", "dev", "smm", "recruit", "рекру"
]
NOT_IT = [
    "продав", "кассир", "бухгалтер", "продаж", "ресепшен", "магазин", "кладовщ",
    "водител", "повар", "маркетолог", "маркетплейс", "юрист", "адвокат",
    r"\bhr\b", "врач", "недвижим", "девелоп",
    "генеральный директор", "исполнительный директор", "конструктор", "претензион",
]

title = s["title"].str.lower()
is_it = title.str.contains("|".join(IT_WORDS)) & ~title.str.contains("|".join(NOT_IT))

it = s[is_it].copy()
print(f"IT: {len(it)} из {len(s)}")
```

```python
is_gross = it["salary_gross"] == True

it["salary_net"] = np.where(is_gross, it["target_salary"] * 0.87, it["target_salary"])

print("медиана до:   ", it["target_salary"].median())
print("медиана после:", it["salary_net"].median())
```

```python
display(it["experience"].value_counts(dropna=False))

by_exp = it.groupby("experience")["salary_net"].agg(["count", "median"])
display(by_exp.sort_values("median"))
```

```python
def work_mode(text):
    text = str(text or "")
    if text == "удалённо":
        return "remote"
    if "удал" in text or "гибрид" in text:
        return "hybrid"
    return "office"


it["remote"] = it["work_format"].map(work_mode)
display(it.groupby("remote")["salary_net"].agg(["count", "median"]).sort_values("median"))
```

```python
import re

pd.set_option("display.float_format", "{:.3f}".format)


def grade(title):
    t = title.lower()
    if re.search(r"руковод|head|директор|cto|team ?lead|тимлид|\blead\b|лид\b", t):
        return "lead"
    if re.search(r"senior|сеньор|ведущ|старш|главн", t):
        return "senior"
    if re.search(r"middle|мидл", t):
        return "middle"
    if re.search(r"junior|джун|младш|стаж|intern|начинающ|помощник", t):
        return "junior"
    return "unknown"


it["grade"] = it["title"].map(grade)
display(it.groupby("grade")["salary_net"].agg(["count", "median"]).sort_values("median"))

it["skills_text"] = it["skills"].map(lambda x: "|".join(x) if x is not None and len(x) else "")
it["description"] = it["description"].fillna("")
it[["title", "skills_text"]].head()
```

```python
from sklearn.base import clone
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.dummy import DummyRegressor
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import Ridge
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder


def evaluate(name, estimator, X, y):
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    scores = cross_validate(clone(estimator), X, y, cv=cv,
                            scoring={"mae": "neg_mean_absolute_error",
                                     "mape": "neg_mean_absolute_percentage_error"})
    mae = -scores["test_mae"].mean()
    mape = -scores["test_mape"].mean() * 100
    print(f"{name:<28} ошибка {mae:>8,.0f} ₽   в процентах {mape:5.1f}%")


def split_skills(text):
    return [s for s in text.split("|") if s]


def make_model(cats, skills=False, desc=False, alpha=3.0):
    parts = [
        ("title", TfidfVectorizer(ngram_range=(1, 2), min_df=2), "title"),
        ("cats", OneHotEncoder(handle_unknown="ignore"), cats),
    ]
    if skills:
        parts.append(("skills", TfidfVectorizer(analyzer=split_skills, min_df=2), "skills_text"))
    if desc:
        parts.append(("desc", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                                              max_features=20000, sublinear_tf=True), "description"))
    pipe = Pipeline([("features", ColumnTransformer(parts)), ("ridge", Ridge(alpha=alpha))])
    return TransformedTargetRegressor(regressor=pipe, func=np.log1p, inverse_func=np.expm1)


X = it[["title", "grade", "experience", "remote", "skills_text", "description"]]
y = it["salary_net"]

evaluate("бейзлайн (медиана)", DummyRegressor(strategy="median"), X, y)
evaluate("название + грейд", make_model(["grade"]), X, y)
evaluate("+ опыт + формат", make_model(["grade", "experience", "remote"]), X, y)
evaluate("+ навыки", make_model(["grade", "experience", "remote"], skills=True), X, y)
evaluate("+ описание", make_model(["grade", "experience", "remote"], skills=True, desc=True), X, y)
```

```python
for alpha in [0.3, 1, 3, 10]:
    evaluate(f"всё, alpha={alpha}",
             make_model(["grade", "experience", "remote"], skills=True, desc=True, alpha=alpha), X, y)
```

```python
final = make_model(["grade", "experience", "remote"], skills=True).fit(X, y)

pipe = final.regressor_
names = pipe.named_steps["features"].get_feature_names_out()
coef = pd.Series(pipe.named_steps["ridge"].coef_, index=names)

print("Сильнее всего ПОВЫШАЮТ зарплату:")
display(coef.sort_values(ascending=False).head(15))
print("Сильнее всего ПОНИЖАЮТ:")
display(coef.sort_values().head(15))
```
