```python
import pandas as pd
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.0f}".format)

df = pd.read_parquet('../data/vacancies.parquet')

print(df.shape)
print(df.columns)
df.head()
```

```python
has_salary = df["salary_from"].notna() | df['salary_to'].notna()

s = df[has_salary]
print(len(s))
```

```python
print(s["currency"].value_counts(dropna=False))
print(s["salary_period"].value_counts(dropna=False))
```

```python
is_rub = s["currency"] == "RUR"
is_month = s["salary_period"] == "month"

s = s[is_rub & is_month]
print(len(s))
```

```python
target_salary = s[['salary_from', 'salary_to']]
s['target_salary'] = target_salary.mean(axis=1)
s['target_salary'].describe()
```

```python
s[['title', 'salary_from', 'salary_to', 'target_salary']].sort_values(ascending=False, by='target_salary').head(10)
```

```python
s[['title', 'salary_from', 'salary_to', 'target_salary']].sort_values(ascending=True, by='target_salary').head(10)
```
